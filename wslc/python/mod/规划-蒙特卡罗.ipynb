#### 以求解圆周率为例
#### 向正方形内投射粒子，判断落在内切圆中的概率

```python
import numpy as np
import matplotlib.pyplot as plt

p = 10000 # 投放的点数
r = 1 # 半径
x,y = 1,1 # 圆心
n = 0 # 内切圆中的点数

plt.figure()
plt.title("Mc")
plt.xlabel("x")
plt.ylabel("y")

for i in range(p):
    px = np.random.rand() * 2 # rand() 返回[0,1)
    py = np.random.rand() * 2
    dis = (px - x)**2 + (py - y)**2
    if dis < r**2:
        plt.plot(px,py,'.',color='blue')
        n += 1
    else :
        plt.plot(px,py,'.',color='red')

plt.axis("equal")

ans = (n / p) * 4 # 由统计规律 p / S方 = n / S圆 解得 pi = ans 
print(f"The calculate pi is {ans}")

plt.show()
```
