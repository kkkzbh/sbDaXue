
#### 在层次分析法中,首先需要判断出影响决策的因素
#### 首先统计因素值，整理成矩阵并相同因素归一化
#### 然后就是要判断这些因素的权值
#### 构造正负反矩阵，即首先主观构造一个相对量，接着根据相对量主观分别表示出$i$相对于$j$的重要程度值$A_{ij}$
#### 然后就是下面的操作

## 1.矩阵处理

```python
import numpy as np

A = np.array([[1,2,3,5],[1/2,1,1/2,2],[1/3,2,1,2],[1/5,1/2,1/2,1]])    # 输入构造的正相反矩阵
n = A.shape[0] # 用shape[0 or 1] 获取行和列的大小信息
eig_val, eig_vec = np.linalg.eig(A) # 得到特征值和特征向量  其中返回的eig_vec是按列对应的特征向量
max_eigv = eig_val.max() # 得到特征值的最大值 这里以调用内置函数形式 也可以用max()

CI = (max_eigv - n) / (n - 1)  # 计算一致性检测值 CI = (λ_max - n) / (n - 1)
RI = [0,0.0001,0.52,0.89,1.12,1.26,1.36,1.41,1.46,1.49,1.52,1.54,1.56,1.58,1.59] # 标准表 CI在对应的值的10%内(不包括)合格

CR = CI / RI[n - 1] # CR的值若在0.1内 则合格

print(f"eig_val = {eig_val}")
print(f"CI = {CI}")
print(f"RI = {RI[n - 1]}")
print(f"CR = {CR}")
```

## 2.特征法求权重


```python


index = np.argmax(eig_val) # argmax() 返回最大值所在的索引
vec = eig_vec[:,index] # 拿到对应的特征向量
weights = vec / sum(vec)  # 归一化,拿到权值向量 这里显式调用sum() 也可以用类内置的.sum()

print(f"weights = {weights}")
```
