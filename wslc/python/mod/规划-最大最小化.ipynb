## 例

$
    \max~\{~\min \left\{ 2x_1 + 3x_2,4x_1 - x_2 \right\}~\}
$
<br>
<br>
$
s.t. \begin{cases}
            x_1,x_2 \geq 0
     \end{cases}
$

## 上述可以等价转化，引入一个量t,使得求解变为
$
    \max ~ t
$
<br><br>
$
s.t. \begin{cases}
        2x_1 + 3x_2 \geq t, \\
        4x_1 - x_2 \geq t, \\
        x_1,x_2 \geq 0
     \end{cases}

$

```python
# 以下以x0 表示 t

import numpy as np
from scipy.optimize import linprog 

# 目标函数的系数向量
c = [-1, 0, 0]

# 约束
A_ub = [[1,-2,-3],
        [1,-4,1],]

B_ub = [0,0]

Bounds = [(None,None),(0,None),(0,None)]

res = linprog(c, A_ub, B_ub, bounds = Bounds, method = "highs")
print(res)
print("最优解:",res.x)
print("最优值:",-res.fun)
```

## 例

![相关数据](规划-最大最小化01.png)
<br><br>
$
        \min_{(x,y)} \left\{~ \max_i \{~ |x - a_i| + |y - bi|~ \} \right\}
$
<br><br>
$
s.t. \begin{cases}
        3 \leq x \leq 8, \\
        4 \leq y \leq 10
     \end{cases}
$


```python
import numpy as np
from scipy.optimize import minimize

a = np.array([1, 4, 3, 5, 9, 12, 6, 20, 17, 8])
b = np.array([2, 10, 8, 18, 1, 4, 5, 10, 8, 9])

# 定义目标函数
def objective(x):
        ret = abs(x[0] - a[0]) + abs(x[1] - b[0])
        for i in range(1,10):
                ret = max(abs(x[0] - a[i]) + abs(x[1] - b[i]),ret)
        return ret

# 随便给出一个初始预测
x0 = np.array([7,7])

bounds = [(3,8),
          [4,10]]

res = minimize(objective, x0, bounds = bounds, method = 'SLSQP')

print("最优解:",res.x)
print("最优值:",res.fun)

```
