## 例
$x:~~~~1~~~~2~~~~3~~~~4~~~~~~5 $
<br>  
$y:~~~~1~~~~2~~~~1.3~~3.75~~2.25$

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 创建数据
x = np.array([[1], [2], [3], [4], [5]]) # 由于可以处理多元线性回归，故这里套的二维数组
y = np.array([1, 2, 1.3, 3.75, 2.25])

# 创建模型并拟合
model = LinearRegression()
model.fit(x,y)

# 获取回归系数和截距
k = model.coef_
b = model.intercept_

# 预测
y_pred = model.predict(x)   # predict 接受一个二维数组 返回一个一维数组的计算值

# 绘图
plt.scatter(x, y, color = "blue") # 数据点
plt.plot(x, y_pred, color = "red") # 回归直线
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression")
plt.show()
```
