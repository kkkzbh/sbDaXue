## 例
$
minimize~ f(x) = x_0^2 + x_1^2 
$
<br>
$
subject~ to~ \begin{cases}
                x_0 + x_1 - 1 \leq 0, \\
                x_0^2 + x_1^2 - 0.5 \leq 0, \\
                -1 \leq x_0,x_1 \leq 1
           \end{cases}
$

```python

import time
import numpy as np
from scipy.optimize import minimize, Bounds

# 定义目标函数
def objective(x):
    return x[0]**2 + x[1]**2

#定义约束
def req1(x):
    return x[0] + x[1] - 1

def req2(x):
    return x[0]**2 + x[1]**2 - 0.5
    
# 定义边界
bounds = Bounds([-1,-1], [1,1])    

# 使用蒙塔卡罗确定初值
np.random.seed(int(time.time()))
ix = np.random.uniform(-1,1,size = 2)

# 定义约束字典
req = ({'type': 'ineq', 'fun': req1},
       {'type': 'ineq', 'fun': req2},)

# 使用内点法解决(默认)
res = minimize(objective, ix,bounds = bounds,constraints = req)

print("最优解:",res.x)
print("最优值",res.fun)

```
