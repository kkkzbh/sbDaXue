### SIR模型
$
\begin{cases}
	\frac{di}{dt} = \lambda si - \mu i , i(0) = i_0 \\
	\frac{ds}{dt} = -\lambda si, s(0) = s_0
\end{cases}
$

$\\lambda = 1, \mu = 0.3, i(0) = 0.02,s(0) = 0.98$

求出$i(t),s(t)$的数值计算结果,$i(t),s(t)$图形 $i \sim s$图形


```python


import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

lam = 1
mu = 0.3
i0 = 0.02
s0 = 0.98

def model(__y,__t,__lambda,__mu):
    i,s = __y;
    di_dt = __lambda * s * i - mu * i;
    ds_dt = -__lambda * s * i;
    return [di_dt,ds_dt]

t = np.linspace(0,100,100)

ini = [i0,s0]

ans = odeint(model,ini,t,args = (lam,mu))

iv = ans[:,0]
sv = ans[:,1]

print(f'iv = {iv}')
print(f'sv = {sv}')

plt.rc('font',family = 'SimHei')
plt.rc('font',size = 16)

plt.plot(t,iv,color = 'green',label = 'i')
plt.plot(t,sv,color = 'red',label = 's')
plt.xlabel('$t$(时间)')
plt.ylabel('$s$ 与 $i$')
plt.xticks(np.arange(0,100 + 20,10))
plt.yticks(np.arange(0,1 + 0.1,0.1))

plt.grid()
plt.legend()
plt.show()

plt.figure()
plt.xlabel('$s$')
plt.ylabel('$i$')
plt.plot(sv,iv,marker = '*',label = r'$i \sim s$')

plt.grid()
plt.legend()
plt.show();

```
